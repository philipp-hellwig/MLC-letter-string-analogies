```python
import warnings
import pandas as pd

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("tables/matching_head_occurrence.csv")
df["train_regime"] = df["copy tasks"].apply(lambda x: "With copy tasks" if x else "Without copy tasks")
df["matching_head"] = df["Matching head"].apply(lambda x: "Present" if x else "Absent")
```

```python
print(f"Matching head occurence without copy tasks: {df[~df["copy tasks"]]["Matching head"].mean()}")
print(f"Matching head occurence with copy tasks: {df[df["copy tasks"]]["Matching head"].mean()}")
```

```python
warnings.filterwarnings('ignore')
plt.style.use("./figures_stylesheet.mplstyle")
my_colors = ["#28479bff"] 

fig, ax = plt.subplots(1, 1, figsize=(3, 2.3))
sns.barplot(df, x="train_regime", y="Matching head", errorbar=None, ax=ax, width=0.6, palette=my_colors)
ax.set_xlabel("Training")
ax.set_ylabel("Proportion matching head present")
plt.savefig("../figures/matching_head_occurrence.svg", format="svg", transparent=True)
```
